```python
import torchaudio
import matplotlib.pyplot as plt
import soundfile as sf
import torch
%matplotlib inline

ds = torchaudio.datasets.LJSPEECH(root = "data")
```

```python
w, sr = sf.read("data/LJSpeech-1.1/wavs/LJ001-0001.wav", dtype="float32")
w = torch.from_numpy(w)
```

```python
fig, ax = plt.subplots(2, 1, figsize=(12, 6))
ax[0].plot(w, lw=0.5)
ax[0].set_title(f"whole clip ({len(w)/sr:.1f} s)")
ax[1].plot(w[20000:20000 + 441])   # 441 samples = 20 ms at 22.05 kHz
ax[1].set_title("20 ms zoom")
plt.tight_layout()
```
